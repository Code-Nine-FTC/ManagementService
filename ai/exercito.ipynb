# Analise de dados simples

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Evolução do estoque ao longo do tempo para cada item
plt.figure(figsize=(12,6))
for item_id in data['item_id'].unique():
    item_data = data[data['item_id'] == item_id]
    plt.plot(item_data['year_month'], item_data['stock_quantity'], marker='o', label=f'Item {item_id}')
plt.xlabel('Ano/Mês')
plt.ylabel('Estoque')
plt.title('Evolução do Estoque por Item')
plt.legend()
plt.show()

# 2. Pedidos realizados por mês (todos os itens)
plt.figure(figsize=(10,5))
sns.barplot(x='year_month', y='orders_placed', data=data, estimator=sum, ci=None)
plt.title('Total de Pedidos Realizados por Mês')
plt.xlabel('Ano/Mês')
plt.ylabel('Pedidos Realizados')
plt.xticks(rotation=45)
plt.show()

# 3. Consumo médio por item
plt.figure(figsize=(10,5))
sns.barplot(x='item_id', y='average_consumed', data=data)
plt.title('Consumo Médio por Item')
plt.xlabel('ID do Item')
plt.ylabel('Consumo Médio')
plt.show()
```
